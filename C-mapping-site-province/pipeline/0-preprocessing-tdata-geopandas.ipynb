Now lets define all the landuse landcover classes in the datasets in a list.
> old_categories = ['Cropland', 'Forest', 'Grassland', 'Mangroves', 'Settlement', 'Water','Bare_Other']

```python
old_categories = ['Cropland', 'Forest', 'Grassland', 'Mangroves', 'Settlement', 'Water','Bare_Other']
```

Lets assign a code to each class based on the number of classes available in a list
> new_groups = [1, 2, 3, 4, 5, 6, 7]

```python
new_groups = [1, 2, 3, 4, 5, 6, 7]
```

Create a mapping dictionary and store the class categories and the class code in it
> category_mapping = dict(zip(old_categories, new_groups))

```python
category_mapping = dict(zip(old_categories, new_groups))
```

Apply mapping 
>tdata['LULC_code'] = tdata['Class'].map(category_mapping)

```python
tdata['LULC_code'] = tdata['Class'].map(category_mapping)
```

Now lets print the class code and the number of points for each claases that is tagged to the value code. 
This particular line works by creating a new column and call it 'LULC_CODE' and set the type to integer since it will hold
value code for each classes.
>tdata['LULC_code'] = tdata['LULC_code'].astype('Int64')

```python
tdata['LULC_code'] = tdata['LULC_code'].astype('Int64')
```
